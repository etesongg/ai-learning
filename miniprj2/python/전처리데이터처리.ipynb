```python
전처리데이터 = r'D:\미니프로젝트\실습데이터\전처리결과\전처리데이터.csv'
격자위경도 = r'D:\미니프로젝트\실습데이터\전처리결과\격자_위경도.csv'
결과파일 = r'D:\미니프로젝트\실습데이터\전처리결과\김해_전처리데이터셋.csv'
```

```python
import pandas as pd

df = pd.read_csv(f'{전처리데이터}', header=0, encoding='cp949', engine='python')
df2 = pd.read_csv(f'{격자위경도}', header=0, encoding='cp949', engine='python')

display(df.head(2))
display(df2.head(2))
```

```python
df['행정기관'] = df['행정기관'] + df['도서관']
del df['도서관']
```

```python
df.rename(columns = {'행정기관':'관공서'},inplace = True)
df.rename(columns = {'주차구획수_sum':'주차장'},inplace = True)
df.rename(columns = {'전기차와충전소_전처리_결과_전기차':'전기차'},inplace = True)
df.rename(columns = {'전기차와충전소_전처리_결과_충전소':'충전소'},inplace = True)
df.rename(columns = {'김해_차량평균속도_avg_speed':'차량속도'},inplace = True)

display(df.head(2))
```

```python
df = df.merge(df2, on='gid')
display(df.head(2))
```

```python
df.to_csv(f'{결과파일}', index=False, sep=',', encoding='cp949')
```
