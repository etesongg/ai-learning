```python
충전소정보 = r'C:\miniprj\실습데이터\전처리결과\충전소정보_격자.csv'
결과파일 = r'C:\miniprj\실습데이터\전처리결과\김해_충전량현황종합_전처리.csv'
```

```python
import pandas as pd

# CSV 불러오기
df = pd.read_csv(f'{충전소정보}', encoding='cp949')
```

```python
# 1~9월 충전량 컬럼과 이용수(횟수) 컬럼 목록
충전량_cols = [f'2023년 0{i}월 충전량' for i in range(1, 10)]
이용수_cols = [f'2023년 0{i}월 충전횟수' for i in range(1, 10)]
```

```python
# 총합 계산
df['총충전량'] = df[충전량_cols].sum(axis=1)
df['총이용수'] = df[이용수_cols].sum(axis=1)

# 일수 계산 (2023년 1~9월은 평년 기준)
월별_일수 = [31, 28, 31, 30, 31, 30, 31, 31, 30]  # 2023년은 윤년 아님
총일수 = sum(월별_일수)

# 일평균 계산
df['일평균충전량'] = df['총충전량'] / 총일수
df['일평균이용수'] = df['총이용수'] / 총일수
```
